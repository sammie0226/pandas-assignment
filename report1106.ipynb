## 2. loc vs iloc

```python
# 1. loc으로 인덱스 10~15 행의 sex, age 열을 선택해 출력하세요.

print(titanic.loc[10:15, "sex":"age"],"\n")


# 2. iloc으로 같은 열(sex는 2번째, age는 3번째 위치)을 행 위치 10:15로 선택해 출력하세요.

print(titanic.iloc[10:15, 2:4],"\n")


# 3. 두 결과의 행 수를 각각 출력하고, 왜 다른지 주석으로 쓰세요.

print(len(titanic.loc[10:15, "sex":"age"]))
print(len(titanic.iloc[10:15, 2:4]))


# 다른 이유 : iloc 는 인덱스의 위치로 값을 찾는 방법으로 기존의 인덱스 슬라이싱의 룰(?)을 따르지만, loc는 행과 열의 이름으로 값을 찾기 때문에 기존의 인덱스의 룰이 아닌 그냥 보이는 대로의 즉, 명시적인 대로 인식해서 실제 인덱스 그대로 10:15 의 값을 찾음.
# => 즉, 인덱스 슬라이싱의 끝값의 기준이 다르다
```

## 4. 조건 마스크

```python
# 1. 나이가 10살 미만인지 검사하는 마스크를 만들어 head(5)를 출력하세요.

mask = titanic["age"] < 10
print(titanic[mask].head(5),"\n")


# 2. 마스크로 10살 미만 승객 수를 출력하세요.

print(len(titanic[mask]))
```

## 5. 복합 조건

```python
# 행 수 출력.

# 1. 3등석 그리고 남성 그리고 생존한 승객 (&)

mask1 = (titanic["pclass"] == 3) & (titanic["sex"] == "male") & (titanic["alive"] == "yes")
print(len(titanic[mask1]))


# 2. 1등석 또는 요금 100 이상인 승객 (|)

mask2 = (titanic["pclass"] == 1) | (titanic["fare"] >= 100 )
print(len(titanic[mask2]))


# 3. 혼자 탑승하지 않은 승객 (~, alone 열 사용)

mask3 = ~(titanic["alone"] == True)
print(len(titanic[mask3]))
```

## 6. isin / between

```python
# 1. isin으로 Cherbourg 또는 Queenstown에서 탑승한 승객을 골라 shape를 출력하세요.

mask = titanic["embark_town"].isin( ["Cherbourg","Queenstown"] )
print(titanic[mask].shape,"\n")


# 2. between으로 나이 20~30세(양 끝 포함) 승객을 골라 shape와 나이의 min·max를 출력하세요.

mask = titanic["age"].between(20,30)       # between default : 양 끝값 포함 (inclusive = "both")
print(titanic[mask].shape)

age = titanic[mask]["age"]
print(age.min())
print(age.max(),"\n")


# 3.은 범위를 양 끝 미포함(inclusive="neither")으로 바꿔 shape를 출력하세요.

mask = titanic["age"].between(20,30,inclusive = "neither")
print(titanic[mask].shape)
```

## 7. loc[조건, 열]

```python
# 12살 이하 그리고 3등석 승객의 age, sex, survived 열만 kids 변수에 담고, 행 수와 생존률(survived 평균, 소수 둘째 자리)을 출력하세요.

kids = titanic.loc[(titanic["age"] <=12) & (titanic["pclass"]==3), ["age","sex","survived"]]
print(len(kids))
#print(kids)
#rate1 = sum(kids["survived"]) / len(kids["survived"]) * 100
#print(type(rate1))
#print(round(rate1,2))
rate = (kids["survived"].mean() * 100).round(2)
print(rate,"%")
#print(type(rate))
```

## 8. query

```python
# 1. 변수 min_fare = 50을 만들고, query로 Cherbourg 탑승 그리고 요금이 min_fare 이상인 승객의 shape를 출력하세요.

min_fare = 50
a = titanic.query(" embark_town == 'Cherbourg' and fare >= @min_fare ")
print(a.shape)


# 2. 5-1번 조건을 query로 다시 써서 행 수가 같은지 확인하세요. (3등석 그리고 남성 그리고 생존한 승객)

x = titanic.query(" pclass == 3 and sex == 'male' and survived == 1 ")
print(len(x))
```

## 9. 정렬

```python
# 1. pclass 오름차순, 같은 등급 안에서는 fare 내림차순으로 정렬해 pclass, fare, sex, age 열의 상위 3행을 출력하세요.

a = titanic.sort_values(
    by = ["pclass", "fare"], 
    ascending=[True, False]
    )
print(a[["pclass","fare","sex","age"]].head(3),"\n")


# 2. 요금 내림차순으로 정렬한 결과를 변수에 담고, iloc[0]과 loc[0]의 fare 값을 각각 출력하세요. 두 값이 다른 이유를 주석으로 쓰세요.

b = titanic.sort_values(by = "fare", ascending=False)
#b1 = titanic.sort_values(by = "fare", ascending=True)
#print(b1)
print(b.iloc[0, 6])
print(b.loc[0, "fare"])


# ★★ 두 값이 다른 이유 : 이건,,,,, 진짜 30분 머리 싸매도 모르겠슴........
```

## 10. 종합 — 20~40세 남성, 1·2등석 vs 3등석

```python
# 질문: "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

# 1. 남성 · 나이 20~40세(between) · 1·2등석(isin) 조건으로 pclass, age, fare, survived 열을 선택하세요.

list = titanic.loc[(titanic["sex"] =="male") & (titanic["age"].between(20,40)) & (titanic["pclass"].isin([1,2])), ["pclass","age","fare","survived"]]
print(list,"\n")


# 2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.

print(len(list))

print((list["survived"].mean() *100).round(2),"%")

high_fare = list.sort_values(by = "fare", ascending = False)
print(high_fare.head(3),"\n")


# 3. 같은 조건을 query로 써서 행 수를 검증하세요.

print(len(titanic.query(" sex == 'male' and 40 >= age >= 20 and (pclass == 1 or pclass == 2) ")),"\n")


# 4. 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.

list_three = titanic.query(" sex == 'male' and 40 >= age >= 20 and pclass == 3 ")
print((list_three["survived"].mean() *100).round(2),"%")

#print((sum(list_three["survived"]) / len(list_three["survived"]))*100)

```

- 일단, 동일한 조건에서 1등석, 2등석, 3등석 각각의 생존률을 아래와 같이 구해봤다. 
- 당연하게도 요금이 가장 비싼 1등석의 생존률이 매우 높게 나왔으나, 주목해야 할 점은 2등석의 생존률이 상당히 저조하다는 것이다. 3개의 등급 중 가장 낮은 3등석과 비교했을 때도 현저히 낮은 것을 볼 수 있다. 
- 이를 통해 개인적인 해석을 해보자면, 역시 돈을 많이 쓰면 살아날 확률이 크다는 것과 동시에 돈을 조금 더 쓰던 아니던 결국 생존은 내가 어떤 클래스에 있는지와는 별개의 문제라는 것? 즉, 운이다. 
- (계산한 생존률이 맞는지 모르겠지만.........)

```python
list_one = titanic.query(" sex == 'male' and 40 >= age >= 20 and pclass == 1 ")
rate_one = (list_one["survived"].mean()*100).round(2)
print("1등석 생존룰", rate_one, "%","\n")

list_two = titanic.query(" sex == 'male' and 40 >= age >= 20 and pclass == 2 ")
rate_two = (list_two["survived"].mean()*100).round(2)
print("2등석 생존률", rate_two, "%","\n")

list_three = titanic.query(" sex == 'male' and 40 >= age >= 20 and pclass == 3 ")
rate_three = (list_three["survived"].mean()*100).round(2)
print("3등석 생존률", rate_three, "%")
```
